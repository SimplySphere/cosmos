# RNA to Protein Translation Steps ( Computer instructions)


>   1. Build a copy of the translation wheel ( dictionary in python)
>   2. Read in DNA : transcribe to RNA ( change all 'T' to 'U')

>   3. Read RNA sequence until we find start code ('AUG'). Mark position.

>   4. Starting with start codon, read 3-letter group from RNA (called a codon)
>   5. For each codon, look up the amino acid in the dictionary.
>   6. Add amino acid to end of amino acid sequence (string) we are building.

>   7. Repeat steps 4-6 until we read a stop codon: 'UAG', 'UAA', or 'UGA'.
>   8. Print out  **amino acid** sequence in format required..... we're DONE !!







## Building the translation dictionary

> In the cell below, we have a python function which builds a dictionary from this file.
> It actually builds two dictionaries:

```
          RNA_to_acid : key = codon, value= one letter amino acid designator
          acid_name:    key = AA designator, value = list with abbrevation and full name of AA
```

````python
path = '/content/drive/MyDrive/Central Dogma/RNA_codons.txt'

with open(path, 'r') as file:
    codon_data = file.read().strip().split("\n")
````

````python
def build_codon_dict():
    """
    Function creates the RNA codon to amino acid and
    acid name maps (dictionaries) from text file
    """
    RNA_to_acid = {}
    acid_name   = {}
    for line in codon_data:
        # print(line)
        codon, abv, letter, name = line.strip().split()
        RNA_to_acid[codon] = letter

        if letter == 'X' or letter in acid_name:
            continue
        else:
            acid_name[letter] = [abv, name]
    return RNA_to_acid, acid_name

RNA_to_acid, acid_name = build_codon_dict()

print(RNA_to_acid)
````

````python
##############################################
#  Program reads in RNA codon dictionary and
#  DNA sequence. Then converts DNA sequence to RNA
#  and translates an RNA sequence into
#  a sequence of amino acids
##############################################

def build_codon_dict():
    """
    Function creates the RNA codon to amino acid and
    acid name maps (dictionaries) from text file
    """
    RNA_to_acid = {}
    acid_name   = {}
    for line in codon_data:
        codon,abv,letter,name = line.strip().split()
        RNA_to_acid[codon] = letter
        if letter == 'X' or letter in acid_name:
            continue
        else:
            acid_name[letter] = [abv, name]
    return RNA_to_acid, acid_name

def DNA_to_RNA(DNA_pattern):
    """
    Function translate DNA sequence to RNA sequence
    by replacing nucleotide 'T' with 'U'.
    Returns RNA string
    """
    RNA_string = DNA_pattern.replace('T','U')
    return RNA_string
#
# Create RNA codon dictionary
#
RNA_to_acid, acid_name = build_codon_dict()
#
# DNA_in = 'GAGCAGCTCTACATGGACCAGTTTCAGCTTTAAAAGCACAAA'

DNA_in = 'TGAACGTTTTAGTGAGGTCCTCGGACTGTTTGGTAGGCGGATTGCTCTGCCTGCCTGGCGGGAAGACGACCAAACTGTAGCGTTTAGAGGAAGTAAAAGTCGTAACAAGGTTTCCGTAGGTGAACCTGCGGAAGGATCAT\
TATCGGTTGCTCGTCTTTACGGACTTGCTCACCCCGGTTAACCCCCCTCGGGACTCGGACGGCTCTCCGCTGTCCGTCCTCGGAGCCTCGACCCGTCTGTCGATGCTCACGACGTCTCCGCCCTGGACGGCTCGATCTTA\
TGTCGGCTGTCGTAGGCTGTGACGGCCCGCGTTTGTTGAGGTACCGTACCCCTTCTGTCGTGTGTGCGACAGGTYGTGGGATGTTTACAACGGTTCAAAGAGCTCGCGCCTCAACCCGGGTTTGTGATCTCCCCCCCTGT\
CGGGATGTGGAGCCCATTTTTGACCATACTTGTCTTTCTGTTCTTACAGCGAGACAGTCTCCGGACGCGAGTCTGAGAATTTAACGATACAACACTATACAGTGGATCATTTGGCTCTGGGGTCGATGAAGAACGGAGCT\
AACTCCGTGATGCCGTGTGAACTGCAGG'
#
# Convert DNA sequence to RNA sequence
RNA_in = DNA_to_RNA(DNA_in)
#
#  Find start codon ('AUG') in RNA sequence
#
str_pos = -1  # index for start codon, stays -1 if not found

################################################
#  YOUR CODE GOES HERE:  Segment #1
#    Use for loop to parse thru the RNA_in sequence
#    letter by letter until 3 letter start code 'AUG' is found.
#
#    Assign index of position of start code to variable
#    named : str_pos (used below)
#
#    If no starting code is found (str_pos = -1), then
#    print message: 'No starting code found in sequence
#    and exit program using sys.exit.
################################################

if 'AUG' in RNA_in:
  str_pos = RNA_in.find("AUG")
else:
  print('No starting code found in sequence')
  import sys
  sys.exit

RNA_seq = RNA_in[str_pos:]    #  Cuts off RNA before start code.
#
#  Lookup sequence of amino acids
#
acid_string = ''   # Empty string to start building AA sequence
#
################################################
#  YOUR CODE GOES HERE: Segment #2
#    Use for loop to go thru the new RNA_seq :

#  1.  Look up each 3 letter codon in the RNA_seq in
#    RNA_to_acid dictionary to get amino acid letter.
#  2.  Add acid letter to a growing amino acid string variable
#      called : acid_string (see above)
#  3. Stop when you reach stop codon(indicate by'X' dictionary value) .
################################################
#
#  Print sequence of amino acids
#

for i in range(len(RNA_seq)//3):
  # print(acid_string)
  acid = RNA_to_acid[RNA_seq[i*3] + RNA_seq[i*3+1] + RNA_seq[i*3+2]]
  if acid == "X":
    break
  acid_string += acid

print ('\n\nDNA sequence is:', DNA_in)
print ('RNA sequence is:', RNA_in,'\n')
print (f'Amino acid letter sequence for this RNA sequence is: {acid_string}')
print ('Amino acid short names in this RNA sequence are: \n')
print ('    ', end="")
abbreviation_string = []
for letter in acid_string:
    abbreviation_string += [acid_name[letter][0]]
print("-".join(abbreviation_string), end="")
print( '\n\n')
````
