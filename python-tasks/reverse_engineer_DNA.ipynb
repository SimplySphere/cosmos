# Reverse Engineering the Genetic Code

The ”Central Dogma” of molecular biology states that sequence information can be trans
ferred between nucleic acids(DNA/RNA) , and from nucleic acids(RNA) to proteins, but
sequence information cannot be transferred among proteins, or from proteins to nucleic
acids. In other words, DNA makes RNA makes protein.

Molecular biologists have been able to isolate and sequence the amino acids of many
common anti-biotics, such as Tyrocidine B1, an anti-biotic produced by a bacteria strain (
Bacillus brevus). Tyrocindine B1 can be used to kill other bacteria like Staphylococcus in
humans.


So, given a peptide sequence (an ordered list of amino acids) for Tyrocindine B1, can
we figure out where in the bacteria genome (i.e. which gene) is responsible for making the
sequence. Then we might target genetic changes to alter or improve the effectiveness of the
anti-biotic.

The goal then is to re-engineer the genetic code for a protein sequence:

> Given a specific protein (peptide sequence), can we figure out the DNA it comes from?

## Example of the problem to solve

For Tyrocindine B1, the amino acid sequence is:

> 3-letter: Val-Lys-Leu-Phe-Pro-Trp-Phe-Asn-Gln-Tyr

> 1-letter: V K L F P W F N Q Y

However, when we examine the genetic code in our ”wheel”, a given amino acid (say L for
leucine) has multiple possible RNA codons that code for it (’CUU’,’CUC’,’CUA’,’CUG’).

So for this sequence of amino acids, we are going to obtain a large number of possible DNA
strings that could produce it. Our task is enumerate/list all possible DNA sequences and
obtain a count.

The sequence above (VKLFPWFNQY) will generate multiple 30-letter DNA sequences. How would you find them
all with a python program?

## Possible python program strategy

> 1. Create a reverse dictionary for the genetic code: Given an amino acid (say L for
leucine), provide a list of all codons that code for it (’CUU’,’CUC’,’CUA’,’CUG’ for
L)
    
> 2. Starting from the first amino acid in sequence, add codons to a growing list of DNA
strings.

> 3. Print out list of DNA strings(candidates) and a total count of DNA strings produced

````python
from google.colab import drive
drive.mount('/content/drive')

path = '/content/drive/MyDrive/Central Dogma/RNA_codons.txt'

with open(path, 'r') as file:
    codon_data = file.read().strip().split("\n")
````

````python
def build_codon_dict():
  acid_to_RNA = {}
  for line in codon_data:
    codon, abv, letter, name = line.strip().split()
    if letter in acid_to_RNA:
      acid_to_RNA[letter] += 1
    else:
      acid_to_RNA[letter] = 1
  return acid_to_RNA
acid_to_RNA = build_codon_dict()
print(acid_to_RNA)
````

````python
peptide = 'MA'         #  Should have 4 DNA sequences

peptide = 'MASS'       #  Should have 144 DNA sequences

peptide = 'VKLFPWFNQY' #  Tyrocindine B1 sequence: Tell me what you get for number of DNAs
````

````python
sequences = 1
for acid in peptide:
  sequences *= acid_to_RNA[acid]
print(sequences)
````
